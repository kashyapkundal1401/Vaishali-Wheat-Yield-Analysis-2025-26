```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Financial Metrics Setup
# Revenue = Yield * Price
data = {
    'Metric': ['Target Yield', 'Actual Yield', 'Target Revenue (MSP)', 'Actual Revenue (Market)'],
    'Value': [180, 105, 436500, 199500]
}
df_finance = pd.DataFrame(data)

# 2. Cost Breakdown (Based on your 6.25-acre Vaishali data)
costs = {
    'Item': ['Seeds', 'Fertilizer', 'Irrigation', 'Labor', 'Diesel (Conventional)', 'Maintenance'],
    'Amount': [7800, 15625, 12500, 35000, 28500, 8000]
}
df_costs = pd.DataFrame(costs)

# 3. Key Calculations
total_investment = df_costs['Amount'].sum()
actual_revenue = 199500
net_profit = actual_revenue - total_investment
roi = (net_profit / total_investment) * 100

print(f"Data Loaded Successfully.")
print(f"Total Investment: ₹{total_investment:,}")
print(f"Net Profit: ₹{net_profit:,}")
print(f"ROI: {roi:.2f}%")
```

def predict_my_profit(yield_goal, sale_price, method='Conventional'):
    """
    Predicts profit based on yield, price, and tillage method.
    Conventional Diesel: ₹28,500 | Zero Tillage Diesel: ₹14,300
    """
    diesel = 28500 if method == 'Conventional' else 14300
    # Other costs remain fixed based on your data
    fixed_costs = 7800 + 15625 + 12500 + 35000 + 8000
    total_costs = fixed_costs + diesel
    
    projected_revenue = yield_goal * sale_price
    projected_profit = projected_revenue - total_costs
